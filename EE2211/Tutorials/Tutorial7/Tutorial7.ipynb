```python
import numpy as np
import matplotlib.pyplot as plt
import sklearn as sk
```

```python
#Question 1
X = np.array([[0.3510, 1.1796, -0.9852], [2.1812, 2.1068, 1.3766], [0.2415, 1.7753, -1.3244], [-0.1096, 1.2747, -0.6316], [0.1544, 2.0851, -0.8320]])
y = np.array([0.2758, 1.4392, -0.4611, 0.6154, 1.0006])

r_1 = np.corrcoef(X[:, 0], y)
r_2 = np.corrcoef(X[:, 1], y)
r_3 = np.corrcoef(X[:, 2], y)

print("Correlation coefficient between X1 and y:", r_1[0, 1])
print("Correlation coefficient between X2 and y:", r_2[0, 1])
print("Correlation coefficient between X3 and y:", r_3[0, 1])
```

```python
#Question 2

x_train = np.array([-10, -8, -3, -1, 2, 7])
y_train = np.array([4.18, 2.42, 0.22, 0.12, 0.25, 3.09])

x_test = np.array([-9, -7, -5, -4, -2, 1, 4, 5, 6, 9])
y_test = np.array([3, 1.81, 0.80, 0.25, -0.19, 0.4, 1.24, 1.68, 2.32, 5.05])

poly_1 = sk.preprocessing.PolynomialFeatures(degree=1)
x_train_poly_1 = poly_1.fit_transform(x_train.reshape(-1, 1))
x_test_poly_1 = poly_1.transform(x_test.reshape(-1, 1))
w_1 = np.linalg.inv(x_train_poly_1.T @ x_train_poly_1) @ x_train_poly_1.T @ y_train
y_pred_1 = x_test_poly_1 @ w_1
MSE_1 = np.mean((y_test - y_pred_1) ** 2)
print("MSE for degree 1 polynomial regression:", MSE_1)

poly_2 = sk.preprocessing.PolynomialFeatures(degree=2)
x_train_poly_2 = poly_2.fit_transform(x_train.reshape(-1, 1))
x_test_poly_2 = poly_2.transform(x_test.reshape(-1, 1))
w_2 = np.linalg.inv(x_train_poly_2.T @ x_train_poly_2) @ x_train_poly_2.T @ y_train
y_pred_2 = x_test_poly_2 @ w_2
MSE_2 = np.mean((y_test - y_pred_2) ** 2)
print("MSE for degree 2 polynomial regression:", MSE_2)

poly_3 = sk.preprocessing.PolynomialFeatures(degree=3)
x_train_poly_3 = poly_3.fit_transform(x_train.reshape(-1, 1))
x_test_poly_3 = poly_3.transform(x_test.reshape(-1, 1)) 
w_3 = np.linalg.inv(x_train_poly_3.T @ x_train_poly_3) @ x_train_poly_3.T @ y_train
y_pred_3 = x_test_poly_3 @ w_3
MSE_3 = np.mean((y_test - y_pred_3) ** 2)
print("MSE for degree 3 polynomial regression:", MSE_3)

poly_4 = sk.preprocessing.PolynomialFeatures(degree=4)
x_train_poly_4 = poly_4.fit_transform(x_train.reshape(-1, 1))
x_test_poly_4 = poly_4.transform(x_test.reshape(-1, 1))
w_4 = np.linalg.inv(x_train_poly_4.T @ x_train_poly_4) @ x_train_poly_4.T @ y_train
y_pred_4 = x_test_poly_4 @ w_4
MSE_4 = np.mean((y_test - y_pred_4) ** 2)
print("MSE for degree 4 polynomial regression:", MSE_4)

poly_5 = sk.preprocessing.PolynomialFeatures(degree=5)
x_train_poly_5 = poly_5.fit_transform(x_train.reshape(-1, 1))
x_test_poly_5 = poly_5.transform(x_test.reshape(-1, 1))
w_5 = np.linalg.inv(x_train_poly_5.T @ x_train_poly_5) @ x_train_poly_5.T @ y_train
y_pred_5 = x_test_poly_5 @ w_5
MSE_5 = np.mean((y_test - y_pred_5) ** 2)
print("MSE for degree 5 polynomial regression:", MSE_5)

poly_6 = sk.preprocessing.PolynomialFeatures(degree=6)
x_train_poly_6 = poly_6.fit_transform(x_train.reshape(-1, 1))
x_test_poly_6 = poly_6.transform(x_test.reshape(-1, 1))
w_6 = x_train_poly_6.T @ np.linalg.inv(x_train_poly_6 @ x_train_poly_6.T) @ y_train
y_pred_6 = x_test_poly_6 @ w_6
MSE_6 = np.mean((y_test - y_pred_6) ** 2)
print("MSE for degree 6 polynomial regression:", MSE_6)

plt.scatter(x_train, y_train, color='blue', label='Training Data')
plt.scatter(x_test, y_test, color='red', label='Test Data')
plt.plot(x_test, y_pred_1, color='green', label='Degree 1 Fit')
plt.plot(x_test, y_pred_2, color='orange', label='Degree 2 Fit')
plt.plot(x_test, y_pred_3, color='purple', label='Degree 3 Fit')
plt.plot(x_test, y_pred_4, color='brown', label='Degree 4 Fit')
plt.plot(x_test, y_pred_5, color='pink', label='Degree 5 Fit')
plt.plot(x_test, y_pred_6, color='gray', label='Degree 6 Fit')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Polynomial Regression Fits')
plt.legend()
plt.show()
```

```python
#Question 2 (b): same as above but with regularization (lambda = 1)

w_1 = np.linalg.inv(x_train_poly_1.T @ x_train_poly_1 + np.eye(x_train_poly_1.shape[1])) @ x_train_poly_1.T @ y_train
y_pred_1 = x_test_poly_1 @ w_1  
MSE_1 = np.mean((y_test - y_pred_1) ** 2)
print("MSE for degree 1 polynomial regression with regularization:", MSE_1)

w_2 = np.linalg.inv(x_train_poly_2.T @ x_train_poly_2 + np.eye(x_train_poly_2.shape[1])) @ x_train_poly_2.T @ y_train
y_pred_2 = x_test_poly_2 @ w_2
MSE_2 = np.mean((y_test - y_pred_2) ** 2)
print("MSE for degree 2 polynomial regression with regularization:", MSE_2)

w_3 = np.linalg.inv(x_train_poly_3.T @ x_train_poly_3 + np.eye(x_train_poly_3.shape[1])) @ x_train_poly_3.T @ y_train
y_pred_3 = x_test_poly_3 @ w_3
MSE_3 = np.mean((y_test - y_pred_3) ** 2)
print("MSE for degree 3 polynomial regression with regularization:", MSE_3)

w_4 = np.linalg.inv(x_train_poly_4.T @ x_train_poly_4 + np.eye(x_train_poly_4.shape[1])) @ x_train_poly_4.T @ y_train
y_pred_4 = x_test_poly_4 @ w_4
MSE_4 = np.mean((y_test - y_pred_4) ** 2)
print("MSE for degree 4 polynomial regression with regularization:", MSE_4)

w_5 = np.linalg.inv(x_train_poly_5.T @ x_train_poly_5 + np.eye(x_train_poly_5.shape[1])) @ x_train_poly_5.T @ y_train
y_pred_5 = x_test_poly_5 @ w_5 
MSE_5 = np.mean((y_test - y_pred_5) ** 2)
print("MSE for degree 5 polynomial regression with regularization:", MSE_5)

#w_6 needs to be computed using dual form since the number of features is greater than the number of training samples
w_6 = x_train_poly_6.T @ np.linalg.inv(x_train_poly_6 @ x_train_poly_6.T + np.eye(x_train_poly_6.shape[0])) @ y_train
y_pred_6 = x_test_poly_6 @ w_6
MSE_6 = np.mean((y_test - y_pred_6) ** 2)
print("MSE for degree 6 polynomial regression with regularization:", MSE_6)
```
