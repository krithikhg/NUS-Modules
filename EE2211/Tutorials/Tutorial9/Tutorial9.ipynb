```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import sklearn as sk
```

```python
#Question 3:

housing = sk.datasets.fetch_california_housing()
housing_df = pd.DataFrame(housing.data, columns=housing.feature_names)
X = housing_df["MedInc"].values
y = housing.target

regressor = sk.tree.DecisionTreeRegressor(criterion='squared_error', max_depth=2)
regressor.fit(X.reshape(-1,1), y)
regressor_pred  = regressor.predict(X.reshape(-1,1))

plt.scatter(X, y, color='blue', label='Actual')
plt.scatter(X, regressor_pred, color='red', label='Predicted')
plt.xlabel('MedInc')
plt.ylabel('House Value')
plt.title('Decision Tree Regression')
plt.legend()
plt.show()

plt.figure(figsize=(18, 8))
sk.tree.plot_tree(
    regressor,
    feature_names=['MedInc'],
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title('Decision Tree Structure (max_depth=2)')
plt.tight_layout()
plt.show()
```

```python
iris = sk.datasets.load_iris()
X = iris.data
y = iris.target

# Split into training and testing sets

X_train, X_test, y_train, y_test = sk.model_selection.train_test_split(X, y, test_size=0.2, random_state=0)

classifier = sk.tree.DecisionTreeClassifier(criterion='entropy', max_depth=4)

accuracy = sk.metrics.accuracy_score(y_test, classifier.fit(X_train, y_train).predict(X_test))
print(f"Accuracy: {accuracy:.2f}")

plt.figure(figsize=(18, 8))
sk.tree.plot_tree(
    classifier,
    feature_names=iris.feature_names,
    class_names=iris.target_names,
    filled=True,
    rounded=True,
    fontsize=6
)
plt.title('Decision Tree Structure (max_depth=4)')
```
